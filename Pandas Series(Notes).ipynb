### What is Pandas

Pandas is an open-source Python library used for data manipulation, data analysis, and data cleaning. It provides powerful data structures such as Series and DataFrame to work with structured data efficiently.

### Pandas Series

A pandas Series is like a column in a table. It is a 1-D array holding data of any type.

### Importing Pandas

```python
import pandas as pd
```

### Series from lists

#### string

```python
country = ['India', 'Nepal', 'Pakistan', 'USA', 'Srilanka']
```

```python
pd.Series(country)
```

#### integers

```python
runs = [13,24,56,78,100]
```

```python
runs_ser = pd.Series(runs)
runs_ser
```

#### custom index

```python
marks = [67,57,89,100]
subjects = ['maths', 'english', 'science', 'hindi']
```

```python
pd.Series(marks, index=subjects)
```

#### setting a name

```python
marks = pd.Series(marks, index=subjects, name = 'Annual marks')
marks
```

### Series from dict

```python
marks = {
    'maths' : 67,
    'english' : 57,
    'science' : 89,
    'hindi' : 100
}
```

```python
marks_series = pd.Series(marks, name='annual marks')
marks_series
```
