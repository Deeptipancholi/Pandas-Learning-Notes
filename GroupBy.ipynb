### agg method

```python
movies.groupby('Genre').agg({
    'Runtime' : 'mean',
    'IMDB_Rating' : 'mean',
    'No_of_Votes': 'sum',
    'Gross' : 'sum',
    'Metascore' : 'min'
})
```

```python
movies.groupby('Genre')[['Runtime','IMDB_Rating','No_of_Votes','Gross','Metascore']].agg(['min','max','mean'])
```

```python
#  Find the highest rated movie of each genre
movies.sort_values('IMDB_Rating', ascending=False).drop_duplicates('Genre')[['Series_Title','IMDB_Rating','Genre','Director']]
```

### apply

```python
# find number of movies starting with A for each group/genres

movies.groupby('Genre')['Series_Title'].apply(lambda x: x.str.startswith('A').sum())
```

```python
# find ranking of each movie in the group according to IMDB score
```

```python
movies['rank'] = movies.groupby('Genre')['IMDB_Rating'].rank(ascending=False)
movies
```

```python
# find normalized IMDB rating wise
```

```python
movies['normalized IMDB rating'] = movies.groupby('Genre')['IMDB_Rating'].apply(lambda x: (x - x.min())/(x.max() - x.min())).droplevel(0)
movies
```

### gorupby on multiple cols

```python
duo = movies.groupby(['Director','Star1'])
duo.size()
duo.get_group(('Aamir Khan','Amole Gupte'))
```

```python
# find the most earning actor and director combo
movies.groupby(['Director','Star1'])['Gross'].sum().sort_values(ascending=False).head(1)
```

```python
# find the number of movies created by specific actor and director

def combo(Director, Star1):
    return movies[(movies['Director'] == Director) & (movies['Star1'] == Star1)].size

combo('Akira Kurosawa','Toshirô Mifune')
combo('Frank Darabont','Tim Robbins')
```

```python
# find the best(in-term of metascore(avg)) actor-> genre combo
movies.groupby(['Star1', 'Genre'])['Metascore'].mean().reset_index().sort_values('Metascore', ascending=False).head(1)
```

```python
#  agg on multiple groupby
movies.groupby(['Director','Star1'])[['IMDB_Rating','No_of_Votes','Gross']].agg(['min','max','mean'])
```

### Excercise

```python
ipl = pd.read_csv('deliveries.csv')
ipl.head()
```

```python
# find the top 10 batsman in terms of runs
ipl.groupby('batsman')['batsman_runs'].sum().sort_values(ascending= False).head(10)
```

```python
# find the batsman with max no of sixes
ipl[ipl['batsman_runs'] == 6]['batsman'].value_counts().sort_values(ascending=False).head(1).index[0]
```

```python
# find batsman with most number of 4's and 6's in last 5 overs
temp_df = ipl[ipl['over'] > 15]
temp_df = temp_df[(temp_df['batsman_runs'] == 4) | (temp_df['batsman_runs'] == 6)]
temp_df.groupby('batsman')['batsman'].count().sort_values(ascending=False).head(1).index[0]
```

```python
ipl[(ipl['over'] > 15) &((ipl['batsman_runs'] == 4) | (ipl['batsman_runs'] == 6))]['batsman'].value_counts().head(1).index[0]
```

```python
# find V Kohli's record aganinst all teams
temp_df = ipl[ipl['batsman'] == 'V Kohli']
temp_df.groupby('bowling_team')['batsman_runs'].sum().reset_index()
```

```python
# create a function that can return the highest score of any batsman
def highest_score(batsman):
    temp_df = ipl[ipl['batsman'] == batsman]
    return temp_df.groupby(['match_id'])['batsman_runs'].sum().max()
 
highest_score('V Kohli')
highest_score('MS Dhoni')
```
